```python
#MIN MAX for chess
def minimax(values, maximizing):
    if maximizing:
        return max(values)
    else:
        return min(values)

move_A = minimax([3, 5], False)
move_B = minimax([2, 8], False)
move_C = minimax([4, 6], False)

print("Move A:", move_A)
print("Move B:", move_B)
print("Move C:", move_C)

best_move = max(move_A, move_B, move_C)

if best_move == move_A:
    print("Best Move: A")
elif best_move == move_B:
    print("Best Move: B")
else:
    print("Best Move: C")

print("Best Score:", best_move)
```

```python
def alpha_beta(values, depth, index, maximizing, alpha, beta):

    # Leaf node
    if depth == 0:
        return values[index]

    if maximizing:
        best = -999

        for i in range(2):
            value = alpha_beta(values, depth - 1,
                               index * 2 + i,
                               False, alpha, beta)

            best = max(best, value)
            alpha = max(alpha, best)

            if beta <= alpha:
                break

        return best

    else:
        best = 999

        for i in range(2):
            value = alpha_beta(values, depth - 1,
                               index * 2 + i,
                               True, alpha, beta)

            best = min(best, value)
            beta = min(beta, best)

            if beta <= alpha:
                break

        return best


# Enter leaf scores
values = list(map(int, input("Enter 8 evaluation scores: ").split()))

result = alpha_beta(values, 3, 0, True, -999, 999)

print("Best Score:", result)
```

```python
#ALPHA BETA PRUNING
def alpha_beta(values, depth, index, maximizing, alpha, beta):

    # Leaf node
    if depth == 0:
        return values[index]

    if maximizing:
        best = -999

        for i in range(2):
            value = alpha_beta(values, depth - 1,
                               index * 2 + i,
                               False, alpha, beta)

            best = max(best, value)
            alpha = max(alpha, best)

            if alpha >= beta:
                break

        return best

    else:
        best = 999

        for i in range(2):
            value = alpha_beta(values, depth - 1,
                               index * 2 + i,
                               True, alpha, beta)

            best = min(best, value)
            beta = min(beta, best)

            if alpha >= beta:
                break

        return best


# Input
values = list(map(int, input("Enter 8 evaluation scores: ").split()))

# Alpha-Beta Pruning
result = alpha_beta(values, 3, 0, True, -999, 999)

print("Best Score:", result)
```

```python
#MIN MAX FOR TIC TAC TOE
def minimax(board, depth, is_max):

    # Check for winner
    if "XXX" in board:
        return 10

    if "OOO" in board:
        return -10

    # Draw
    if " " not in board:
        return 0

    if is_max:
        best = -999

        for i in range(9):
            if board[i] == " ":
                board[i] = "X"
                score = minimax(board, depth + 1, False)
                board[i] = " "
                best = max(best, score)

        return best

    else:
        best = 999

        for i in range(9):
            if board[i] == " ":
                board[i] = "O"
                score = minimax(board, depth + 1, True)
                board[i] = " "
                best = min(best, score)

        return best


# Input board
print("Enter Tic-Tac-Toe board:")
board = list(input("Enter 9 values (X/O/-): "))

# Convert - to empty space
for i in range(9):
    if board[i] == "-":
        board[i] = " "

best_score = -999
best_move = -1

# Find best move for X
for i in range(9):
    if board[i] == " ":
        board[i] = "X"
        score = minimax(board, 0, False)
        board[i] = " "

        if score > best_score:
            best_score = score
            best_move = i + 1

print("Best Move:", best_move)
print("Best Score:", best_score)
```
